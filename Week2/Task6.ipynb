# Week 2: Exploratory Data Analysis (EDA) - Titanic Dataset

## Objective
To analyze the Titanic dataset and extract meaningful insights using data visualization and preprocessing techniques

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = sns.load_dataset("titanic")
df.head()
```

## Dataset Overview

The dataset contains information about Titanic passengers such as age, gender, class, fare, and survival status

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['age'].fillna(df['age'].median(), inplace=True)
df['embarked'].fillna(df['embarked'].mode()[0], inplace=True)
df.drop(columns=['deck'], inplace=True)
```

### Insight
Missing values were handled using median (for age) and mode (for categorical data).

```python
sns.countplot(x='survived', data=df)
plt.title("Survival Count")
plt.show()
```

### Insight
More passengers died than survived.

```python
sns.countplot(x='sex', hue='survived', data=df)
plt.title("Gender vs Survival")
plt.show()
```

### Insight
Females had a much higher survival rate than males.

```python
sns.countplot(x='class', hue='survived', data=df)
plt.title("Class vs Survival")
plt.show()
```

### Insight
Passengers in first class had higher survival rates.

```python
sns.histplot(df['age'], bins=30, kde=True)
plt.title("Age Distribution")
plt.show()
```

### Insight
Most passengers were between 20–40 years old.

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

## Key Observations & Insights

- Females had a significantly higher survival rate.
- First-class passengers were more likely to survive.
- Younger passengers had slightly better survival chances.
- Fare is positively correlated with survival (higher fare → higher survival probability).

## Conclusion

EDA helps us understand patterns in data. In this dataset, survival was influenced by gender, class, and fare. Visualization makes it easier to interpret these relationships.
