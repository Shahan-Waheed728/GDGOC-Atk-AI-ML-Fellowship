# Task1 : Handling Missing Data – Titanic Dataset


**Problem:** *The Titanic dataset contains missing values in columns like Age, Cabin,* *and Embarked.*

**Technique:** *Mean/Median imputation for numerical, Mode imputation for categorical,* *Drop for excessive missing.*

**Why:** *Missing data causes errors in ML models and biases predictions.*

```python
#Data Loading & Preprocessing

import pandas as pd
import numpy as np

df = pd.read_csv('/kaggle/input/datasets/shuofxz/titanic-machine-learning-from-disaster/train.csv')

# --- Before ---
print("Missing Values BEFORE")
print(df.isnull().sum())

# Strategy 1: Mean imputation for Age (numerical)
df['Age_mean'] = df['Age'].fillna(df['Age'].mean())

# Strategy 2: Median imputation for Age (robust to outliers)
df['Age_median'] = df['Age'].fillna(df['Age'].median())

# Strategy 3: Mode imputation for Embarked (categorical)
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Strategy 4: Drop Cabin column (too many missing values >77%)
df.drop(columns=['Cabin'], inplace=True)

# Apply median to Age finally (best strategy)
df['Age'] = df['Age_median']
df.drop(columns=['Age_mean', 'Age_median'], inplace=True)

# --- After ---
print("\nMissing Values AFTER")
print(df.isnull().sum())

# ## 3. Results & Observations
print("\n=== Observations ===")
print(f"Age - Mean: {df['Age'].mean():.2f}, Median: {df['Age'].median():.2f}")
print("Cabin dropped due to 77% + missing values.")
print("Embarked filled with mode: 'S'")
```

**Results and Observations:**

**What changed?** *Age nulls filled, Embarked nulls filled, Cabin dropped*

**Before vs After?** *Age had 177 missing → 0, Cabin had 687 missing → removed,* *Embarked had 2 missing → 0*

**What did we learn?** *Mean is sensitive to outliers; median is safer for skewed data like Age. Columns with >50% missing are better dropped than imputed.*

# Task2: Feature Encoding – Car Evaluation Dataset

**Introduction**

**Problem:** *ML models can't process raw categorical text values.*

**Technique:** *Label Encoding & One-Hot Encoding compared.*

**Why:** *Encoding converts categories to numbers models can understand.*

```python
# Data Loading & Preprocessing

import pandas as pd
from sklearn.preprocessing import LabelEncoder

cols = ['buying', 'maint', 'doors', 'persons', 'lug_boot', 'safety', 'class']
df = pd.read_csv('/kaggle/input/datasets/rohit265/car-evaluation-uci-machine-learning/car.csv', names=cols)

print("Before Encoding")
print(df.head())
print(df.dtypes)

# --- Label Encoding ---
df_label = df.copy()
le = LabelEncoder()
for col in df_label.columns:
    df_label[col] = le.fit_transform(df_label[col])

print("\nAfter Label Encoding")
print(df_label.head())

# --- One-Hot Encoding ---
df_ohe = pd.get_dummies(df, columns=['buying', 'maint', 'doors', 'persons', 'lug_boot', 'safety'])

print("\nAfter One-Hot Encoding")
print(df_ohe.head())
print(f"Shape after OHE: {df_ohe.shape}")
print(f"Shape after Label Encoding: {df_label.shape}")
```

**Results and observations:**

**Label Encoding:** *stays same shape, assigns integer ranks — can imply false order* *(e.g. vhigh=3 > low=1)*

**One-Hot Encoding:** *expands columns, no false ordering, but increases dimensionality*

**When to use which?** *Label encoding for ordinal data (low < med < high), OHE for* *nominal (no order)*

**Introduction**

**Problem:** *Features have different ranges, biasing ML models.*

# Task 4: Handling Outliers – Boston Housing Dataset

**Introduction**

**Problem:** *Outliers skew model learning and degrade performance.*

**Technique:** *Z-score and IQR method to detect and remove outliers.*

**Why:** *Clean data leads to more accurate and generalizable models.*

```python
# Data Loading & Preprocessing

import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv('/kaggle/input/datasets/subho117/boston-housing-kaggle-challenge-with-lr/boston.csv')
df.fillna(df.median(), inplace=True)

print(f"Shape Before: {df.shape}")

# Z-score method
z_scores = np.abs(stats.zscore(df))
df_zscore = df[(z_scores < 3).all(axis=1)]
print(f"Shape After Z-score removal: {df_zscore.shape}")

# IQR method
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1
df_iqr = df[~((df < (Q1 - 1.5 * IQR)) | (df > (Q3 + 1.5 * IQR))).any(axis=1)]
print(f"Shape After IQR removal: {df_iqr.shape}")
```

**Results & Observations:**

*Z-score removed rows where any feature was 3 std devs away from mean.*

*IQR is more robust for skewed distributions.*

*Both methods reduced dataset size, removing noisy extreme values.*


# Task 5: Advanced Data Imputation – Retail Sales Dataset

**Introduction:**

**Problem:** *Simple imputation ignores relationships between features.*

**Technique:** *KNN Imputation and MICE (IterativeImputer).*

**Why:** *Advanced methods use feature correlations for smarter imputation.*

```python
# Data Loading & Preprocessing

import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

df = pd.read_csv('/kaggle/input/datasets/mohammadtalib786/retail-sales-dataset/retail_sales_dataset.csv')

# Keep only numeric columns
df_numeric = df.select_dtypes(include=[np.number]).copy()

# Introduce artificial missing values for demonstration
np.random.seed(42)
mask = np.random.rand(*df_numeric.shape) < 0.1
df_missing = df_numeric.copy()
df_missing[mask] = np.nan

print(f"Missing before: {df_missing.isnull().sum().sum()}")

# KNN Imputation
knn = KNNImputer(n_neighbors=5)
df_knn = pd.DataFrame(knn.fit_transform(df_missing), columns=df_numeric.columns)

# MICE Imputation
mice = IterativeImputer(max_iter=10, random_state=42)
df_mice = pd.DataFrame(mice.fit_transform(df_missing), columns=df_numeric.columns)

print(f"Missing after KNN: {df_knn.isnull().sum().sum()}")
print(f"Missing after MICE: {df_mice.isnull().sum().sum()}")
```

**Results & Observations**

*KNN fills based on nearest neighbor feature values.*

*MICE iteratively models each feature using others — more accurate.*

*Both eliminate missing values; MICE is preferred for complex datasets.*

#  Task 6: Feature Engineering – Heart Disease Dataset

**Introduction**

**Problem:** *Raw features may not fully capture domain knowledge.*

**Technique:** *Create derived features — age groups, cholesterol categories, risk* *score.*

**Why:** *New features can improve model performance significantly.*

```python
import pandas as pd

df = pd.read_csv('/kaggle/input/datasets/fedesoriano/heart-failure-prediction/heart.csv')

# Fix column names
df.columns = df.columns.str.lower()

print("Before Feature Engineering")
print(df.head())

# Age groups
df['age_group'] = pd.cut(df['age'], bins=[0,40,55,70,100],
                        labels=['Young','Middle','Senior','Elderly'])

# Cholesterol categories
df['chol_category'] = pd.cut(df['cholesterol'], bins=[0,200,240,600],
                            labels=['Normal','Borderline','High'])

# Risk score
df['risk_score'] = df['restingbp'] + df['cholesterol']/5 - df['maxhr']/2

print("\nAfter Feature Engineering")
print(df[['age','age_group','cholesterol','chol_category','risk_score']].head(10))

```

**Results & Observations**

*Age groups convert continuous age into meaningful clinical categories.*

*Cholesterol categories align with medical thresholds.*

*Risk score combines multiple features into one informative signal.*


# Task 7: Variable Transformation – Bike Sharing Dataset

**Introduction**

**Problem:** *Skewed distributions violate ML model assumptions.*

**Technique:** *Log, Square Root, and Box-Cox transformations.*

**Why:** *Transformations normalize skewed data for better model performance.*

```python
# Data Loading & Preprocessing

import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv('/kaggle/input/datasets/sriramm2010/uci-bike-sharing-data/hour.csv')

target = 'cnt'
print(f"Skewness Before: {df[target].skew():.4f}")

# Log Transform
df['cnt_log'] = np.log1p(df[target])

# Square Root Transform
df['cnt_sqrt'] = np.sqrt(df[target])

# Box-Cox Transform (requires positive values)
df['cnt_boxcox'], _ = stats.boxcox(df[target] + 1)

print(f"Skewness Log: {df['cnt_log'].skew():.4f}")
print(f"Skewness Sqrt: {df['cnt_sqrt'].skew():.4f}")
print(f"Skewness BoxCox: {df['cnt_boxcox'].skew():.4f}")

```

**Results & Observations**

*Original 'cnt' was right-skewed.*

*Log transform reduced skewness significantly.*

*Box-Cox is the most powerful — finds optimal lambda automatically.*

*Sqrt is a mild transformation, useful for moderate skewness.*

# Task 8: Feature Selection – Diabetes Dataset

**Introduction**

**Problem:** *Too many features cause overfitting and slow training.*

**Technique:** *Correlation, Mutual Information, and RFE.*

**Why:** *Selecting relevant features improves model accuracy and efficiency.*

```python
# Data Loading & Preprocessing

import pandas as pd
from sklearn.feature_selection import mutual_info_classif, RFE
from sklearn.linear_model import LogisticRegression

df = pd.read_csv('/kaggle/input/datasets/organizations/uciml/pima-indians-diabetes-database/diabetes.csv')

X = df.drop('Outcome', axis=1)
y = df['Outcome']

# Correlation
print("Correlation with Target")
print(df.corr()['Outcome'].sort_values(ascending=False))

# Mutual Information
mi = mutual_info_classif(X, y, random_state=42)
mi_series = pd.Series(mi, index=X.columns).sort_values(ascending=False)
print("\nMutual Information Scores")
print(mi_series)

# RFE
model = LogisticRegression(max_iter=1000)
rfe = RFE(model, n_features_to_select=5)
rfe.fit(X, y)
selected = X.columns[rfe.support_].tolist()
print(f"\n=== RFE Selected Features ===")
print(selected)

```

**Results & Observations**

*Correlation: Glucose is most correlated with Outcome.*

*Mutual Information: Captures non-linear relationships too.*

*RFE: Recursively eliminates weakest features — selects top 5.*

*All 3 methods agree Glucose, BMI, Age are most important.*

# Task 9: Handling Imbalanced Data – Credit Card Fraud Detection

**Introduction**

**Problem:** *Fraud cases are rare (~0.17%), causing model bias toward majority class.*

**Technique:** *SMOTE, ADASYN, and Undersampling.*

**Why:** *Balancing classes ensures the model learns fraud patterns effectively.*

```python
# Data Loading & Preprocessing

import pandas as pd
from imblearn.over_sampling import SMOTE, ADASYN
from imblearn.under_sampling import RandomUnderSampler
from collections import Counter

df = pd.read_csv('/kaggle/input/datasets/aliiihussain/credit-card-fraud-detection/credit_card_fraud_synthetic.csv')

X = df.drop('Class', axis=1)
y = df['Class']

print(f"Before: {Counter(y)}")

# SMOTE
sm = SMOTE(random_state=42)
X_sm, y_sm = sm.fit_resample(X, y)
print(f"After SMOTE: {Counter(y_sm)}")

# ADASYN
ada = ADASYN(random_state=42)
X_ada, y_ada = ada.fit_resample(X, y)
print(f"After ADASYN: {Counter(y_ada)}")

# Undersampling
rus = RandomUnderSampler(random_state=42)
X_rus, y_rus = rus.fit_resample(X, y)
print(f"After Undersampling: {Counter(y_rus)}")

```

**Results & Observations**

**Original:** *284315 legit vs 492 fraud — heavily imbalanced.*

**SMOTE:** *Synthesizes new fraud samples to balance classes.*

**ADASYN:** *Focuses synthesis on harder-to-classify samples.*

**Undersampling:** *Reduces majority class — risk of losing information.*

*SMOTE preferred when dataset is large enough.*

# Task 11: Dimensionality Reduction – MNIST Dataset

**Introduction**

**Problem:** *MNIST has 784 features (pixels) — high dimensional, slow to train.*

# Task 12: Text Preprocessing – IMDB Movie Reviews

**Introduction**

**Problem:** *Raw text is unstructured and can't be fed directly to ML models.*

**Technique:** *Lowercasing, stopword removal, stemming, lemmatization, tokenization.*

**Why:** *Clean text reduces noise and improves model accuracy.*

```python
#  Data Loading & Preprocessing

import pandas as pd
import re

df = pd.read_csv('/kaggle/input/datasets/vishakhdapat/imdb-movie-reviews/IMDB Dataset.csv')

print("Before")
print(df['review'][0][:300])

# Manual stopwords list (no download needed)
stop_words = set([
    'i','me','my','myself','we','our','ours','ourselves','you','your','yours',
    'yourself','he','him','his','himself','she','her','hers','herself','it',
    'its','itself','they','them','their','theirs','themselves','what','which',
    'who','whom','this','that','these','those','am','is','are','was','were',
    'be','been','being','have','has','had','having','do','does','did','doing',
    'a','an','the','and','but','if','or','because','as','until','while','of',
    'at','by','for','with','about','against','between','into','through',
    'during','before','after','above','below','to','from','up','down','in',
    'out','on','off','over','under','again','further','then','once','here',
    'there','when','where','why','how','all','both','each','few','more',
    'most','other','some','such','no','nor','not','only','own','same','so',
    'than','too','very','s','t','can','will','just','don','should','now'
])

def simple_stem(word):
    # Basic suffix stripping (poor man's stemmer)
    for suffix in ['ing', 'tion', 'ed', 'ly', 'er', 'es', 's']:
        if word.endswith(suffix) and len(word) - len(suffix) > 2:
            return word[:-len(suffix)]
    return word

def preprocess(text):
    text = text.lower()                              # Lowercase
    text = re.sub(r'<.*?>', '', text)               # Remove HTML tags
    text = re.sub(r'[^a-z\s]', '', text)           # Remove punctuation/numbers
    tokens = text.split()                            # Tokenize (split on space)
    tokens = [t for t in tokens if t not in stop_words]  # Remove stopwords
    tokens = [simple_stem(t) for t in tokens]       # Stem
    return ' '.join(tokens)

df['clean_review'] = df['review'].apply(preprocess)

print("\nAfter")
print(df['clean_review'][0][:300])

print(f"\nOriginal avg length: {df['review'].str.len().mean():.0f} chars")
print(f"Cleaned avg length:  {df['clean_review'].str.len().mean():.0f} chars")



```

**Results & Observations**

*HTML tags, punctuation, and noise removed.*

*Stopwords eliminated common but uninformative words.*

*Lemmatization + Stemming reduced words to their root forms.*

**Result:** *compact, clean text ready for vectorization (TF-IDF, embeddings).*

# Task 13: Time-Series Preprocessing – Air Quality Dataset

**Introduction**

**Problem:** *Time-series data has missing timestamps, irregular intervals, and noise.*

**Technique:** *Resampling, forward fill, and rolling mean smoothing.*

**Why:** *Clean, regular time-series data is essential for forecasting models.*


```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    '/kaggle/input/datasets/nishantbhadauria/datasetucimlairquality/AirQualityUCI.csv',
    sep=',',   
)

# Clean column names
df.columns = df.columns.str.strip().str.lower()

# Combine date + time
df['datetime'] = pd.to_datetime(df['date'] + ' ' + df['time'], dayfirst=True)

df.drop(columns=['date', 'time'], inplace=True)
df.set_index('datetime', inplace=True)

# Replace invalid values
df.replace(-200, np.nan, inplace=True)

print(f"Missing before fill: {df.isnull().sum().sum()}")

# Fill missing values
df = df.ffill()

print(f"Missing after fill: {df.isnull().sum().sum()}")

# Resample
df_hourly = df.resample('h').mean(numeric_only=True)

# Smoothing
df_hourly['co_smoothed'] = df_hourly['co_gt'].rolling(window=24, min_periods=1).mean()

print(df_hourly[['co_gt', 'co_smoothed']].head())




```

**Results & Observations**

*-200 sensor errors replaced with NaN then forward-filled.*

*Resampling to hourly intervals creates a uniform time index.*

*Rolling mean smoothing removes noise while preserving trend.*

*Data is now ready for LSTM or ARIMA forecasting models.*
